Introduction : 
Une exoplanète est une planète qui tourne autour d'une étoile qui n'est pas le Soleil
Il faut savoir que lorsqu’une exoplanète passe devant son étoile, elle bloque une petite partie de la lumière. 
Ce passage d’une planète devant son étoile est appelé un transit. 
Si on observe des “creux” réguliers dans une courbe de lumière, cela indique qu'une planète passe devant l’étoile.
Ainsi est obtenue une exoplanète candidate dont l'existence devra être confirmée.
Mon objecrtif est donc de simuler ce phénomène de transit à l'aide de Python et comprendre comment les chercheurs arrivent à confirmer l'existence d'exoplanètes.

La méthode du "folding" consiste à définir un point de départ (l'epoch) qui correspond au tout premier point de la courbe (plus globalement au début d'un cycle). A partir de là, la courbe sera pliée un certain nombre de fois afin de superposer tous les cycles (par exemple tous les transits). La méthode du binning consiste quant à elle à regrouper les points de données en "petits groupes", calculer la moyenne dans chaque groupe afin de lisser la courbe.

A présent, je vais créer mon propre programme basé sur la méthode BLS (Box Least Sqaures) afin d'obtenir la période orbitale d'une planète de manière plus générale. Pour cela, je vais créer une fonction qui calculera la puissance d'une unique période : elle sera nommée "bls_power_one_period". Ensuite, je créerai une deuxilème fonction qui testera toutes les périodes dans un intervalle donnée et créera un tableau : elle sera nommée "bls_scan".

```python
def fold(time, flux, period):
    phase = (time % period) / period
    return phase, flux
```

```python
import numpy as np
def bls_power_one_period(time, flux,period, bins=1000):
    phase, f = fold(time, flux, period) # Je plie la courbe sur la période "period"
    sorted_indices = np.argsort(phase) # Cela permet de créer des indices qui numérotent les données de "phase" afin de pouvoir obtenir des valeurs croissantes qui seront placés dans un tableau
    phase = phase[sorted_indices] 
    f = f[sorted_indices] # J'organise le tableau afin que les points de la courbe soient dans l'ordre croissant 
    bin_means = np.array_split(f, bins) 
    means = [np.mean(b) for b in bin_means] # On calcule la moyenne du flux dans chaqque "bin"
    power = np.max(means) - np.min(means) # On fait la différence entre les valeurs des flux des bins les plus élevés et les moins élevés 
    return power
```

```python
import numpy as np
def bls_scan(time, flux, min_periods=0.2, max_periods=7, n_periods=900) : 
    periods = np.linspace(min_periods, max_periods, n_periods) # On demande au programme BLS de tester 900 périodes qui durent entre 0,2 et 7 jours
    powers = [] # Je crée une liste vide qui stokera chaque puissance calculée ci-dessous
    for p in periods: # Je calcule la puissance de chaque période
        power = bls_power_one_period(time, flux, p) # Toutes les périodes dans le tableau "periods" sont parcourus
        powers.append(power) # Je rajoute la puissance à la liste
    return periods, np.array(powers) # Je convertis la liste avec les puissances calculées en tableau
```

```python
time = np.linspace(0, 10, 1000) 
flux = np.sin(2 * np.pi * time / 2)  # Je simule de manière simplifié le flux par la fonction sinus pour tester ton code
periods, powers = bls_scan(time, flux, min_periods=0.5, max_periods=200, n_periods=600)
print("Périodes testées :", periods)
print("Puissances correspondantes :", powers)
```

```python
import matplotlib.pyplot as plt
best_period = periods[np.argmax(powers)]
print("Période estimée :", best_period, "jours")
plt.plot(periods, powers)
plt.xlabel("Période (jours)")
plt.ylabel("Puissance du transit (approx.)")
plt.title("Périodogramme BLS simplifié")
plt.show()
```

Le résultat obtenu n'est pas forcément celui auquel je m'attendais bien que j'avais conscience qu'un flux simplifié comme ceui ci-dessus n'allait pas aboutir à un périodogramm BLS parfait. C'est pour cela qu'il me faut intégrer un nouveau paramètre à savoir la profondeur du transit. La position exact du transit étant inconnue, on crée donc une "boite" qui se "déplace" progressivement le long de la phase définie ci-dessus. On calcule ensuite la profondeur comme la différence entre le flux moyen hors de la boite et dans la boite.

```python
import numpy as np

def fold(time, flux, period):
    phase = (time % period) / period
    return phase, flux

def bls_power_one_period(time, flux, period, q=0.05, n_t0=400):
    phase, f = fold(time, flux, period)
    sorted_indices = np.argsort(phase)
    phase = phase[sorted_indices]
    f = f[sorted_indices]
    
    best_depth = 0
    for t0 in np.linspace(0, 1 - q, n_t0): # On teste 50 périodes réparties entre 0 et 1-q
        in_box = (phase >= t0) & (phase <= t0 + q)
        out_box = ~in_box # out_box contient tous les points hors du transit (hors de la boite)
        depth = np.mean(f[out_box]) - np.mean(f[in_box])
        if depth > best_depth:
            best_depth = depth

    return best_depth
```

```python
import matplotlib.pyplot as plt
periods = np.linspace(0.5, 5, 400)
powers = np.array([bls_power_one_period(time, flux, p) for p in periods])
best_depth=np.max(powers)
best_period = periods[np.argmax(powers)]
print("Période estimée :", best_period, "jours")
print("Profondeur maximale :", best_depth)
plt.plot(periods, powers)
plt.xlabel("Période (jours)")
plt.ylabel("Puissance du transit (approx.)")
plt.title("Périodogramme BLS")
plt.show()
```

On teste à présent ce programme BLS pour Kepler-10 b. Pour cela, il faut simplement que je remplace les valeurs "time" et "flux" dans la partie du programme "bls_scan".

```python
import numpy as np
def bls_power_one_period(time, flux,period, bins=1000):
    phase, f = fold(time, flux, period) 
    sorted_indices = np.argsort(phase) 
    f = f[sorted_indices] 
    bin_means = np.array_split(f, bins) 
    means = [np.mean(b) for b in bin_means] 
    power = np.max(means) - np.min(means) 
    return power
```

```python
import numpy as np
def bls_scan(time, flux, min_periods=0.2, max_periods=7, n_periods=900) : 
    periods = np.linspace(min_periods, max_periods, n_periods)
    powers = [] 
    for p in periods:
        power = bls_power_one_period(time, flux, p) 
        powers.append(power) 
    return periods, np.array(powers)
```

Pour terminer ce projet, je vais mesurer la profondeur du transit δ (correspondant à la fraction de lumière bloquée par l’exoplanète), le rayon de la planète à l'aide d'une formule contenant la profondeur du transit et la distance orbitale à l'aide de la troisième loi de Kepler.

```python
Fmin=np.min(lc_fold_binned.flux.value) 
Fhors=np.median(lc_fold_binned.flux.value) # On calcule la valeurs moyenne du flux en dehors du transit (Fhors) et sa valeur minimale (Fmin)
depth=(Fhors-Fmin) / Fhors
print("Profondeur normalisée du transit:",depth)
```

On obtient une profondeur du transit de 9,6e-05, c'est-à-dire que Kepler-10 b bloque 0,0096 % de la lumière de l'étoile, ce qui est cohérent étant donné que Kepler-10 b est une petite planète rocheuse.

Le rayon de la planète Kepler-10 b se calcule à l'aide de la formule suivante : 

```python
R_étoile = 776000000 # en m # On utilise la valeur officielle du rayon de Kepler-10
R_planète = R_étoile*np.sqrt(depth)
print("Rayon de la planète:",R_planète)
```

Enfin, on calcule la distance ortibale de Kepler-10 b à l'aide de la troisième loi de Kepler. D'après la troisième loi de Kepler, pour tous les astres en orbite autour du même astre attracteur, le rapport entre le carré de la période de révolution
T et le cube de la longueur 𝑎 du demi-grand axe est égal à une constante qui ne dépend que de l’astre attracteur : $\dfrac{4\pi^2}{GM_{étoile}}$

```python
G = 6.674e-11           # en m^3 kg^-1 s^-2
M_etoile=1.78e30      # en kg
P = best_period*24*3600 # en secondes
UA = 1.496e11  # en mètres
a = (G * M_etoile * P**2 / (4 * np.pi**2))**(1/3) # en mètres
print("La distance orbitale en mètres de Kepler-10 est :", a )
print("La distance orbitale en UA de Kepler-10 est :", a/UA ) # On peut aussi mettre la distance orbitale en unité astronomique
```
